# 1. Background

## 1.1 What is Anaconda?

Anaconda is an open-source **distribution** of the Python and R programming languages, tailored
for scientific computing, data analysis and machine learning. It simplifies package management
and deployment, and ships with a large collection of pre-installed libraries.

Its four principal advantages are:

| Advantage | Why it matters to you |
|---|---|
| Bundles the essential data-science packages (`NumPy`, `pandas`, `matplotlib`, `scikit-learn`) | No manual dependency resolution on day one |
| Includes integrated tools: Jupyter Notebook, Spyder, Anaconda Navigator | One install gives you the whole workbench |
| Provides `conda`, which manages packages **and** isolated environments | Two projects with conflicting requirements can coexist |
| Cross-platform | Windows, macOS and Linux behave identically |

## 1.2 What is Jupyter Notebook?

Jupyter Notebook is a **browser-based interactive development environment** in which Python code
is written and executed in a sequence of *cells*. A notebook interleaves executable code, its
output, and rich documentation (Markdown, equations, plots) in a single file.

This makes it especially suitable for analysis, reporting and teaching, because:

- it supports step-by-step coding and analysis in segmented cells;
- it renders data visualisations inline, next to the code that produced them;
- it encourages reproducibility and transparency in analytical work;
- it is the de facto standard in both academia and industry.

### The three standard inspections

| Method | What it tells you, and why it matters |
|---|---|
| `df.head()` | Displays the first five rows. Confirms the file parsed correctly, the header row was recognised, and the delimiter was right. |

| `df.describe()` | Count, mean, standard deviation, min, max and the 25th / 50th (median) / 75th percentiles for numeric columns — central tendency, spread and skew at a glance. |
| `df.shape` | The pair (rows, columns). Records the size of the problem before any filtering. |

# 3. Git and Version Control

## 3.1 A story of solving problems before they start

Amir, a computer science student, had just started his first major project — a web-based quiz
system. He wrote code every day and made fast progress, but as the project grew so did its
complexity. One day he changed several files to add a login feature. After testing he
discovered that not only was the new feature broken, it had also broken parts of the program
that had previously worked. He tried to undo the changes by hand, but could no longer remember
what the original code looked like.

A friend introduced him to **Git**. With Git, Amir could *commit* changes as he worked, like
saving checkpoints in a game; each commit is a safe, restorable state of the code. If something
went wrong he could return to an earlier version — an unlimited undo button for the entire
project. He then discovered **GitHub**, which stores Git projects online, so that his work was
backed up in the cloud, could be shared with classmates, and could be shown publicly as a
portfolio.

## 3.2 Branches and nodes: exploring without losing the way

As Amir grew more confident he wanted to try new features — a dark mode, a new grading
algorithm — without endangering the working version. This is what **branches** are for.

Think of the project as a tree: the `main` branch is the trunk, the stable working version. A
new branch is a new limb: a separate line of development where changes can be made and tested
freely. Every commit on any branch is stored as a **node** — a snapshot in the history. If the
experiment succeeds, the branch is *merged* back into the trunk; if it fails, the branch is
deleted and the trunk is untouched.

The picture that results is called the **Git graph**:

```
                     F1 ----- F2 ----- F3            feature/login
                    /                     \
  C1 ---- C2 ------                         ------ C5    main
                     \                     /
                       C3 ------ H1 -------           hotfix/typo
```

### Command reference

| Command | Effect |
|---|---|
| `git status` | Shows which files are modified, staged or untracked. Run it before every commit. |
| `git add <path>` | Stages a change for the next commit. |
| `git commit -m "..."` | Creates a new node in the history with the staged changes. |
| `git log --oneline --graph` | Draws the Git graph in the terminal. |
| `git checkout -b <name>` | Creates a branch and switches to it. |
| `git merge <name>` | Merges the named branch into the current one. |
| `git push` / `git pull` | Synchronises the local history with GitHub in each direction. |

# 5. Home Assignment

Write `lab01/inspect.py` containing a function `inspect(path_or_url)` that loads a CSV and prints
a single compact report: shape, dtypes, per-column missing counts and percentages, and the
numeric summary.

Run it against **two different public datasets**, commit the script and both reports on a branch
named `lab01/homework`, open the branch on GitHub, and merge it. Submit the URL of the merge
commit.

A skeleton is provided below — complete it, then save it as `inspect.py`.

````python
def inspect(path_or_url):
    """Load a CSV and print a compact, standard first-look report.

    Parameters
    ----------
    path_or_url : str
        A local file path or an http(s) URL pointing at a CSV file.
    """
    import pandas as pd

    df = pd.read_csv(path_or_url)

    print("=" * 70)
    print("SOURCE :", path_or_url)
    print("SHAPE  :", df.shape[0], "rows x", df.shape[1], "columns")
    print("=" * 70)

    print("\n--- DATA TYPES ---")
    print(df.dtypes)

    print("\n--- MISSING VALUES ---")
    miss = df.isnull().sum()
    miss = miss[miss > 0]
    if miss.empty:
        print("None.")
    else:
        report = pd.DataFrame({
            "missing": miss,
            "percent": (miss / len(df) * 100).round(1),
        })
        print(report.sort_values("missing", ascending=False))

    print("\n--- NUMERIC SUMMARY ---")
    print(df.describe())

    return df


# Test it on the Titanic dataset
_ = inspect("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/titanic.csv")
````
